```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Set display options to see all columns clearly
pd.set_option('display.max_columns', None)

# Load the dataset
df = pd.read_csv('social_media_engagement_5000.csv')
print("Dataset loaded successfully!")
```

```python
# Task 1.1: Check initial data types and convert the date/time column
print("--- Initial Dataset Info ---")
display(df.info())
# Explicitly convert 'posted_at' to datetime format with dayfirst=True
df['posted_at'] = pd.to_datetime(df['posted_at'], dayfirst=True)

print("\nSuccessfully converted 'posted_at' to datetime format!")
display(df[['posted_at']].dtypes)

# Task 1.2: Use NumPy arrays for basic calculations
likes_arr = df['likes'].to_numpy()
impressions_arr = df['impression_count'].to_numpy() # Note: column is 'impression_count'

print("\n--- NumPy Array Operations ---")
print(f"Total Likes across dataset: {np.sum(likes_arr):,}")
print(f"Average Impressions: {np.mean(impressions_arr):,.2f}")
print(f"Max Likes: {np.max(likes_arr):,}")
print(f"Min Likes: {np.min(likes_arr):,}")
```

```python
print("--- Dataset Structure & Shape ---")
print("Shape (Rows, Columns):", df.shape)
print("\nColumns List:")
print(df.columns.tolist())

print("\n--- First 3 Rows Preview ---")
display(df.head(3))
```

```python
print("--- Dataset Info & Data Types ---")
display(df.info())
```

```python
print("--- Summary Statistics for Numerical Fields ---")
display(df.describe())
```

```python
print("--- Post Type Distribution ---")
display(df['post_type'].value_counts())

print("\n--- Unique Count Check ---")
print(f"Unique Countries: {df['country'].nunique()}")
print(f"Unique Sentiments: {df['sentiment'].unique().tolist()}")
```

```python
# Select only numeric columns
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()

print("--- Correlation Matrix ---")
display(corr_matrix)
```

```python
print("--- Average Likes by Post Type ---")
avg_likes_by_type = df.groupby('post_type')['likes'].mean().reset_index()
display(avg_likes_by_type)

print("\n--- Total Impressions by Country ---")
impressions_by_country = df.groupby('country')['impression_count'].sum().reset_index()
display(impressions_by_country)
```

```python
# Create an engagement_score metric (weighted sum: likes*1 + comments*2 + shares*3)
df['engagement_score'] = (df['likes'].fillna(0) * 1) + \
                         (df['comments'].fillna(0) * 2) + \
                         (df['shares'].fillna(0) * 3)

print("--- Engagement Score Created Successfully ---")
display(df[['post_id', 'likes', 'comments', 'shares', 'engagement_score']].head(3))
```

```python
# Multi-level groupby summary by country, post type, and sentiment
multi_group_summary = df.groupby(['country', 'post_type', 'sentiment']).agg({
    'engagement_score': 'mean',
    'impression_count': 'sum',
    'post_id': 'count'
}).rename(columns={'post_id': 'total_posts'}).reset_index()

print("--- Multi-Level Groupby Summary (Top 5 rows) ---")
display(multi_group_summary.head())
```

```python
# Task 5: Compute descriptive stats for core numerical metrics
target_metrics = ['likes', 'comments', 'shares', 'watch_time_sec', 'engagement_rate', 'follower_count']
stats_list = []

for col in target_metrics:
    if col in df.columns:
        s = df[col].dropna()
        stats_list.append({
            'Metric': col,
            'Mean': s.mean(),
            'Median': s.median(),
            'Mode': s.mode()[0] if not s.mode().empty else np.nan,
            'Std Dev': s.std(),
            'Variance': s.var(),
            '90th Percentile': s.quantile(0.90)
        })

stats_df = pd.DataFrame(stats_list)
print("--- Statistical Analysis Summary ---")
display(stats_df)
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(df['impression_count'], df['likes'], alpha=0.3, color='blue')
plt.title('Likes vs Impressions')
plt.xlabel('Impressions')
plt.ylabel('Likes')
plt.show()
```

```python
plt.figure(figsize=(8, 4))
daily_trend = df.groupby('posted_at')['engagement_score'].mean()
plt.plot(daily_trend.index, daily_trend.values, color='green')
plt.title('Daily Engagement Trend')
plt.xlabel('Date')
plt.ylabel('Average Engagement Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 4))
df['post_type'].value_counts().plot(kind='bar', color='orange', edgecolor='black')
plt.title('Posts by Type')
plt.xlabel('Post Type')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(5, 5))
if 'gender' in df.columns:
    df['gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', colors=['skyblue', 'pink', 'lightgray'])
    plt.title('Gender Distribution')
    plt.ylabel('')
    plt.show()
```

```python
plt.figure(figsize=(6, 4))
if 'age' in df.columns:
    plt.hist(df['age'].dropna(), bins=20, color='purple', edgecolor='black')
    plt.title('Age Distribution')
    plt.xlabel('Age')
    plt.ylabel('Frequency')
    plt.show()
```

```python
plt.figure(figsize=(6, 3))
plt.boxplot(df['engagement_rate'].dropna(), orientation='horizontal', patch_artist=True, boxprops=dict(facecolor='lightblue'))
plt.title('Box Plot of Engagement Rate')
plt.xlabel('Engagement Rate')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x='post_type', hue='sentiment', palette='Set2')
plt.title('Sentiment Count by Post Type')
plt.xlabel('Post Type')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(9, 4))
sns.boxplot(data=df, x='country', y='engagement_score', hue='country', palette='Set3', legend=False)
plt.title('Engagement Score Distribution by Country')
plt.xticks(rotation=45)
plt.xlabel('Country')
plt.ylabel('Engagement Score')
plt.show()
```
